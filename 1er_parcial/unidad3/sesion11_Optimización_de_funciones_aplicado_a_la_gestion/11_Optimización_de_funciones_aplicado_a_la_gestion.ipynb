### ¿Qué es un monopolio? 🏭
▶¿Qué es un monopolio?
Un monopolio es una estructura de mercado donde una única empresa es la única proveedora de un bien o servicio que no tiene sustitutos cercanos. Esto le otorga a la empresa un poder significativo sobre el precio y la cantidad ofrecida en el mercado, ya que no enfrenta competencia directa.

image.png

*Características*

* Único productor: Solo hay una empresa que domina todo el mercado.
* Barreras a la entrada: Existen obstáculos que impiden que otras empresas ingresen al mercado, como patentes, control de recursos esenciales,Informacion adicional, altos costos fijos o regulaciones gubernamentales.
* Poder de mercado: El monopolista puede fijar precios por encima del costo marginal, lo que le permite obtener beneficios extraordinarios.
* Ausencia de sustitutos cercanos: El producto o servicio ofrecido no tiene alternativas similares que los consumidores puedan elegir.
* Tambien puede influir cuestiones como los switching cost o efectos de red.

*¿Cómo surge un monopolio?*
* Recursos exclusivos: Cuando una empresa controla un recurso clave necesario para la producción (ejemplo: una mina de diamantes única).
* Economías de escala: Si los costos medios disminuyen significativamente con la producción, una empresa grande puede ser más eficiente que varias pequeñas (ejemplo: servicios públicos como el agua o la electricidad).
* Patentes y licencias: El gobierno puede otorgar derechos exclusivos para producir un bien o servicio (ejemplo: patentes de medicamentos).
* Prácticas anticompetitivas: Una empresa puede eliminar a sus competidores mediante tácticas predatorias, como vender a precios por debajo del costo.

image.png


* * *



**🧩Cada curva nos da una informacion determinada de la estructura del monopolio**

🔵 Demanda (curva azul descendente): muestra el precio que los consumidores están dispuestos a pagar por cada cantidad.

🔴 Ingreso marginal (IM): es decreciente y siempre está por debajo de la demanda (porque para vender una unidad más, el monopolio debe bajar el precio para todos).

🔴 Costo marginal (CM): sube con la cantidad producida (curva en U o creciente).

🔴 Costo total promedio (CTP): muestra el costo por unidad producida en promedio (también en U).

```python
#Funcion Demanda Inversa
def demanda(q):
    return 100 - 2*q

#Funcion Costo total
def CTP(q):
    return 20*q + 10

# Función de beneficio
def beneficio(q):
    return demanda(q) * q - CTP(q)

# Veamos el beneficio cuando llevamos 10 (usando la función numérica correcta)
print("\nBeneficio cuando q = 10:", beneficio(10))
```

# Caso de A: Minimización de Costos en YPF 🛢️

Imaginemos a YPF, una empresa líder en energía, buscando optimizar sus operaciones para minimizar los costos de producción en una de sus plantas extractoras de petróleo.

image.png

La estructura de costos totales de esta planta responde a la siguiente función:

$$C(q) = q^2 - 40q + 500$$

Donde:
*   $C(q)$ representa el costo total de producir $q$ unidades (por ejemplo, barriles de petróleo).
*   $q$ es la cantidad de petróleo producido.
*   El término $q^2 - 40q$ representa los costos variables, que dependen del nivel de producción. Estos costos pueden incluir la mano de obra, la energía consumida en la extracción y el procesamiento, y otros insumos que varían con la cantidad producida. La forma cuadrática ($q^2$) sugiere que puede haber rendimientos decrecientes o crecientes a escala en ciertos rangos de producción, o costos marginales que aumentan con la cantidad.
*   El término $500$ representa los costos fijos. Como mencionaste, estos 500 millones de dólares (o la unidad monetaria correspondiente) son costos que no varían con el nivel de producción en el corto plazo. En el contexto de una planta extractora de petróleo, estos altos costos fijos se deben a la **enorme inversión inicial** en infraestructura, maquinaria especializada (torres de perforación, equipos de bombeo, oleoductos), amortización de activos a largo plazo y el mantenimiento básico de las instalaciones, independientemente de cuánto petróleo se extraiga. Estos son ejemplos de "costos hundidos" o **altos costos de amortización y maquinaria** que son característicos de industrias de capital intensivo como la petrolera.


El objetivo de YPF, desde el punto de vista de la eficiencia operativa de esta planta, es encontrar el nivel de producción ($q$) que minimice este costo total. Esto es crucial para maximizar la rentabilidad general de la empresa.

```python
q = sp.Symbol('q', real=True)

# Función de costo
C = q**2 - 40*q + 500
```

*   Condición de primer orden para un mínimo:

Para encontrar el mínimo de una función, se calcula su primera derivada con respecto a la variable de interés ($q$ en este caso) y se iguala a cero. La derivada de $C(q)$ con respecto a $q$ es:
$$\frac{dC}{dq} = 2q - 40$$
Igualando a cero:
$$2q - 40 = 0$$

*   condición de segundo orden para verificar que es un mínimo:
Para confirmar que el punto encontrado es un mínimo (y no un máximo o un punto de inflexión), se calcula la segunda derivada de la función de costo. Si la segunda derivada es positiva en el punto crítico, se trata de un mínimo.
$$\frac{d^2C}{dq^2} = \frac{d}{dq}(2q - 40) = 2$$



```python
# Derivada primera (condición de mínimo)
C_deriv = sp.diff(C, q)

# Igualamos a cero
q_opt = sp.solve(C_deriv, q)[0]

# Evaluamos el costo mínimo
C_min = C.subs(q, q_opt)

# Derivada segunda para verificar que es mínimo
C_deriv2 = sp.diff(C_deriv, q)

# Resultados
print(f"Función C(q) =")
C
```

Si la demanda es lineal decreciente y el costo es lineal o cuadrático, el ingreso total es una parábola invertida, y el beneficio termina siendo una función cóncava como:

$$π(q)= -aq^2 +bq+c$$
Entonces ya sabemos que tiene un único máximo.

👉Imaginemos un escenario hipotético en el que Coca-Cola es la única marca autorizada para vender bebidas azucaradas en un país. El gobierno, preocupado por la salud pública, prohíbe a otras marcas competir, pero no regula el precio.

Esto convierte a Coca-Cola en un monopolio legal, con total control sobre la oferta y el precio.

*Demanda inversa* : $$P(q)= 50-q $$
*Costo Total* : $$C(q)=10q$$

q representa los millones de litros vendidos al mes de coca-cola. Los consumidores están dispuestos a pagar hasta $50 por litro si se vendiera solo una unidad (demanda muy inelástica al principio, por hábito).

Coca-Cola tiene costos constantes: le cuesta $10 producir y distribuir cada litro (sin costo fijo).

```python
#definimos
def demanda(q):
    return 50 - q
def ingreso(q):
    return demanda(q) * q
def costo(q):
    return 10 * q
def beneficio(q):
    return ingreso(q) - costo(q)

#Maximizamos
res = minimize_scalar(lambda q: -beneficio(q), bounds=(0, 50), method='bounded')
q_opt1 = res.x
p_opt1 = demanda(q_opt1)
beneficio_max = beneficio(q_opt1)

#imprimimos
print(f"Cantidad óptima q* = {q_opt1:.2f}")
print(f"Precio óptimo p* = {p_opt1:.2f}")
print(f"Beneficio máximo = {beneficio_max:.2f}")
```

## ¿como sabemos si un monopolio es bueno para la sociedad?

Para evaluar el impacto social de un monopolio, comparamos el bienestar total en el mercado bajo diferentes estructuras: monopolio vs competencia perfecta.

El indicador principal que usamos es el *excedente del consumidor (EC)*.

💰 ¿Qué es el excedente del consumidor?
Es la diferencia entre lo que los consumidores están dispuestos a pagar y lo que efectivamente pagan. Representa el "ahorro" o satisfacción neta de los consumidores.

📐 ¿Cómo se calcula?
Cuando la demanda es lineal, el excedente del consumidor se calcula como el área de un triángulo: $$EC=\frac{1}{2}(Pmax-P^*)Q^*$$

Donde:

Pmax: es el precio máximo que alguien estaría dispuesto a pagar (intersección con el eje vertical).

P*: el precio de equilibrio (en este caso, el que pone el monopolio).

Q*: la cantidad que se vende en el mercado.





```python
# Excedente del consumidor
EC = 0.5 * (50 - p_opt1) * q_opt1
print(f"Excedente del consumidor = {EC:.2f}")
```

```python
# Gráfico
q_vals = np.linspace(0, 50, 200)
p_vals = demanda(q_vals)

plt.figure(figsize=(8,6))
plt.plot(q_vals, p_vals, label="Demanda", color='blue')

# Precio óptimo horizontal
plt.hlines(p_opt1, 0, q_opt1, colors='green', linestyles='--', label=f'P* = {p_opt1:.2f}')
plt.vlines(q_opt1, 0, p_opt1, colors='red', linestyles='--', label=f'q* = {q_opt1:.2f}')

# Área de excedente del consumidor
plt.fill_between(q_vals[q_vals <= q_opt1], demanda(q_vals[q_vals <= q_opt1]), p_opt1, color='cyan', alpha=0.3, label='Excedente consumidor')

plt.title("Equilibrio de Monopolio y Excedente del Consumidor")
plt.xlabel("Cantidad (q)")
plt.ylabel("Precio (p)")
plt.legend()
plt.grid(True)
plt.show()
```

¿Entonces es bueno para la sociedad? para saberlo debemos sacar el bienestar social para ello debemos sumar el excedente del consumidor mas el beneficio de las empresa/s:

$$Bienestar \ social= Excedente \ consumidor + Beneficio \ Firma$$

```python
B_social=EC+beneficio_max

print(f"El beneficio social es = {B_social:.2f}")
```

🤔Este numero por lo general se contrasta contra un duopolio, triopolio, oligopolio, para analizar que sucederia ante una fusion de empresas, como repercute en la sociedad. Este tipo de analisis, que son aun mas sofisticados recaen en defensa de la competencia que analizan que las firmas potencialmente estrategicas no abusen de su posicion dominante perjudicando a los consumidores

Hemos nombrado que en situaciones de monopolios naturales, pero hay que entender que en la gran mayoria de los casos las innovaciones cientificas y/o patentes vienen por parte de monopolios/oligopolios, vease el ejemplo de las farmaceuticas para indagar.
